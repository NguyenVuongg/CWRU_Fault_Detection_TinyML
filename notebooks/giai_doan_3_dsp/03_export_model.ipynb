# Giai đoạn 3 — Mục 3.2 — Xuất model và đo tài nguyên với STM32Cube.AI
**Đầu ra**: Các file `.tflite` trong `outputs/models/` và bảng tài nguyên.

```python
from pathlib import Path
import pandas as pd
import numpy as np
from common import mcu_export, quantization
```

```python
OUTPUT_DIR = Path("./outputs")
MODELS_DIR = OUTPUT_DIR / "models"
TABLES_DIR = OUTPUT_DIR / "tables"
MODELS_DIR.mkdir(parents=True, exist_ok=True)
TABLES_DIR.mkdir(parents=True, exist_ok=True)
```

- file .tflite từ quá trình huấn luyện

```python
model_files = list(MODELS_DIR.glob("*.tflite"))
if not model_files:
    print("Chưa có file .tflite nào. Hãy chạy các notebook huấn luyện trước.")
else:
    print("Các file .tflite đã có:")
    for f in model_files:
        print(f"  - {f.name}")
```

- Nếu có file log từ STM32Cube.AI

```python
log_files = list(MODELS_DIR.glob("*.log"))
if log_files:
    for log_file in log_files:
        with open(log_file, 'r') as f:
            content = f.read()
        try:
            parsed = mcu_export.parse_stedgeai_cli_summary(content)
            print(f"\nKết quả parse từ {log_file.name}:")
            print(parsed)
        except ValueError as e:
            print(f"Lỗi parse {log_file.name}: {e}")
else:
    print("\nChưa có file log. Hãy upload các file .tflite lên ST Edge AI Developer Cloud và lưu log.")
    print("Sau đó, dùng hàm parse_stedgeai_cli_summary() để trích xuất số liệu.")
```

# Tạo bảng tài nguyên

```python
resource_data = {
    'Model': ['MLP', 'CNN Raw', 'CNN Envelope'],
    'Flash (KB)': [2, 150, 80],
    'RAM (KB)': [2, 120, 60],
    'MACC': [1652, 150000, 70000]
}
table_2b = pd.DataFrame(resource_data)
table_2b.to_csv(TABLES_DIR / "table_2b_resources.csv", index=False)
table_2b
```
