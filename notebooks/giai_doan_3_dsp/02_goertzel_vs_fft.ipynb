# 02 — Goertzel vs FFT
# Giai đoạn 3 — Mục 3.1b — So sánh MACC và RAM
**Đầu ra**: `outputs/tables/table_3b_goertzel_vs_fft.csv`

```python
from pathlib import Path
import numpy as np
import pandas as pd
```

```python
OUTPUT_DIR = Path("./outputs")
TABLES_DIR = OUTPUT_DIR / "tables"
TABLES_DIR.mkdir(parents=True, exist_ok=True)
```

```python
N = 2048
M = 21  # số tần số mục tiêu

macc_goertzel = M * N
macc_fft = N * np.log2(N)

ram_goertzel = N * 4  # chỉ cần mảng float cho tín hiệu
ram_fft = N * 8  # buffer complex

table_3b = pd.DataFrame({
    'Method': ['Goertzel', 'FFT'],
    'MACC': [macc_goertzel, macc_fft],
    'RAM (bytes)': [ram_goertzel, ram_fft]
})
table_3b.to_csv(TABLES_DIR / "table_3b_goertzel_vs_fft.csv", index=False)
table_3b
```
